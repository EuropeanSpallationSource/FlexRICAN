```python
import subprocess
```

```python
process = subprocess.Popen(
    ["python", "scripts/Scenario_4_essri.py"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)
for line in process.stdout:
    print(line, end="")

process.wait()
```
